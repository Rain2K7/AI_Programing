### 1. Các kiểu dữ liệu
Trong Python có các kiểu dữ liệu thông dụng như:
*   Dữ liệu số: `int` (số nguyên), `float` (số thập phân), `complex` (số phức)
*   Dữ liệu nhị phân: `bool` (True/False)
* Dữ liệu dạng sequence: `str` (strings), `list` (danh sách: dãy có trật tự, độ dài, nội dung thay đổi được), `tuple` (bộ: dãy có trật tự, độ dài, nội dung cố định)
* Tập hợp: `set` (tập hợp không có trật tự gồm các phần tử khác nhau, độ dài và nội dung thay đổi được), `frozenset` (như `set`, nhưng độ dài và nội dung cố định)
* Kiểu ánh xạ: `dict` (từ điển: gồm các cặp key-value, không có trật tự)



```python
# Numeric types
int_var = 10
float_var = 10.5
complex_var = 3 + 4j

# Boolean type
bool_var = True

# Sequence types
str_var = "Hello Python"
list_var = [1, 2, 3, "four"]
tuple_var = (10, 20, "thirty")

# Set types
set_var = {1, 2, 3, 2}

# Mapping type
dict_var = {"name": "Alice", "age": 30}

print(f"int_var: {int_var}, type: {type(int_var)}")
print(f"float_var: {float_var}, type: {type(float_var)}")
print(f"complex_var: {complex_var}, type: {type(complex_var)}")
print(f"bool_var: {bool_var}, type: {type(bool_var)}")
print(f"str_var: {str_var}, type: {type(str_var)}")
print(f"list_var: {list_var}, type: {type(list_var)}")
print(f"tuple_var: {tuple_var}, type: {type(tuple_var)}")
print(f"set_var: {set_var}, type: {type(set_var)}")
print(f"dict_var: {dict_var}, type: {type(dict_var)}")
```

Nhờ bản chất kiểu động, 1 biến có thể khai báo không cần kiểu, và cũng có thể gán lại giá trị thuộc kiểu khác.

```python
a = 5
print(f'Type: {type(a)}')
a = 'Five'
print(f'Type: {type(a)}')
a = False
print(f'Type: {type(a)}')
```

**Vì sao lại có sự vượt quá dung lượng (overhead) này?**

**CPython** là bản implementation phổ biến nhất của Python và được viết hầu hết bằng C. Trong CPython, chi tiết cài đặt của các kiểu dữ liệu có thể thay đổi tùy phiên bản. Tuy nhiên, mô hình chung vẫn là:

> **Mọi giá trị trong Python đều là một đối tượng.**

Một biến Python nên được hiểu là một **tên được liên kết** (*bound*) với một đối tượng. Chính đối tượng mang giá trị và kiểu dữ liệu; tên biến không có kiểu cố định riêng.

Khi gán cho biến một giá trị mới, Python chỉ liên kết tên biến ấy với một đối tượng khác. Đối tượng mới có thể thuộc một kiểu hoàn toàn khác.



```python
a = 5
a = 'Five'
```

Có thể hình dung mô hình này như sau:
Khi `a` được gán cho giá trị `5` ta có:

`a` $\to$ PyLongObject
*   header (thông tin quản lí như số tham chiếu đến đối tượng, ...)
* type: `int`
* value: 5

Khi `a` được gán lại cho giá trị `'Five'` ta có:

`a` $\to$ PyUnicodeObject
* header
* type: `str`
* value: `'Five'`

Vậy tự thân `a` không có kiểu, mà chỉ được liên kết tới những đối tượng khác nhau, nên sử dụng tên biến `a` rất linh hoạt. Đổi lại, những metadata như header sẽ thêm overhead cho dung lượng.


Các cấu trúc dữ liệu như `list`, `set`, `tuple`, `str` hay `dict` cũng là đối tượng. Như đã thấy, Python cho phép 1 cấu trúc dữ liệu kiểu `list`, `set`, `tuple`, `dict` có thể chứa nhiều kiểu dữ liệu, vì các phần tử đều chỉ tham chiếu đến các đối tượng rải rác trên heap.
Các phép toán có thể thực hiện trên từng đối tượng như sau:

| Kiểu    | Ý nghĩa                 | Có thể lặp? | Có thể thay đổi? | Phép toán quan trọng                              |
| ------- | ----------------------- | :---------: | :--------------: | ------------------------------------------------- |
| `list`  | Dãy phần tử có thứ tự   |      Có     |        Có        | indexing, slicing, `append`, `pop`, `sort`        |
| `tuple` | Dãy cố định có thứ tự   |      Có     |       Không      | indexing, slicing, unpacking                      |
| `str`   | Dãy ký tự Unicode       |      Có     |       Không      | indexing, slicing, `split`, `join`, `replace`     |
| `set`   | Tập phần tử không lặp   |      Có     |        Có        | `in`, `add`, `remove`, hợp, giao, hiệu            |
| `dict`  | Ánh xạ từ key đến value |      Có     |        Có        | lookup theo key, `get`, `items`, `keys`, `values` |

#### 1.1. Phép lặp

Một object là có thể lặp (**iterable**) nếu ta có thể lấy lần lượt các phần tử của nó. Vì vậy, các cấu trúc trên đều có thể dùng trong vòng lặp `for`:



```python
my_list = [1, 'two', True]
# Iterate by index
for i in range(len(my_list)):
  print(my_list[i])
# Iterate by value
for value in my_list:
  print(value)
```

Python có 2 cách lặp (iterate), đó là thông qua chỉ số (số ít: index, số nhiều: indices), hoặc thông qua từng phần tử trong cấu trúc dữ liệu. Nhưng không phải cấu trúc nào cũng có thể có chỉ số. Chẳng hạn, lặp như sau sẽ báo lỗi:

```python
my_set = {1, 'two', True}
for i in range(len(my_set)):
  print(my_set[i])
```

Vậy phải xem bản chất từng kiểu là gì, mới biết nó có thể lặp bằng chỉ số hay không. Các kiểu `list`, `tuple`, `str` có đặc điểm chung là có thứ tự, nên mới có thể có chỉ số được. Còn `set`, `dict` thì phải lặp cách khác.

```python
for item in my_set:
  print(item)
print(my_set)
```

Khoan! Tại sao `my_set` có 3 phần tử, khi in ra chỉ có 2? Đó là vì kiểu `set` là 1 tập hợp các phần tử phân biệt, mà Python lại xem `1` và `True` là như nhau.

Kiểu `dict` là 1 tập hợp các cặp key-value. Cách tạo và lặp 1 dict:

```python
my_dict = {'one': 1, 'two': 2, 'three': 3}
# Iterate by key
for key in my_dict:
  print(key)
# Iterate by value
for value in my_dict.values():
  print(value)
```

Các kiểu `tuple`, `str` đều có thể lặp bằng chỉ số và phần tử như `list`.

```python
my_tuple = (1, 'two', True)
# Iterate by index
for i in range(len(my_tuple)):
  print(my_tuple[i])
# Iterate by item
for value in my_tuple:
  print(value)
my_string = 'Hello'
# Iterate by index
for i in range(len(my_string)):
  print(my_string[i])
# Iterate by item
for c in my_string:
  print(c)
```

#### 1.2. Phép truy cập giá trị

Để truy cập từng giá trị (value) trong `dict`, ta không dùng chỉ số, mà dùng key.

Nếu key ta cung cấp không tồn tại, sẽ báo lỗi nếu chỉ xài indexing (toán tử []). Nhưng phương thức get() cho ta quyền output ra 1 giá trị khác.

```python
my_dict['zero']
```

```python
print(my_dict.get('zero', 0))
```

Kiểu `set` không cho phép truy cập giá trị tuỳ ý.

Kiểu `list`, `tuple`, `str` thì có thể truy cập bằng chỉ số.

```python
print(my_list[0])
print(my_tuple[1])
print(my_string[2])
print(my_string[len(my_string)])
```

#### 1.3. Phép kiểm tra sự tồn tại
Tất cả các kiểu `list`, `tuple`, `set`, `str`, `dict` đều có thể kiểm tra sự tồn tại của phần tử. Ta dùng từ khoá `in`.

```python
print('two' in my_list)
print('H' in my_string)
print(0 in my_set)
print(True in my_tuple)
```

Riêng kiểu `dict` thì có thể kiểm tra sự tồn tại của 1 key, 1 value, hay cả 1 cặp key-value.

```python
print('one' in my_dict)
print('one' in my_dict.keys())
print(1 in my_dict.values())
print(('one', 1) in my_dict.items())
```

```python
print(my_dict)
print(my_dict.keys()), print(type(my_dict.keys()))
print(my_dict.values()), print(type(my_dict.values()))
print(my_dict.items()), print(type(my_dict.items()))
```

#### 1.4. Chỉnh sửa cấu trúc dữ liệu


Phép chỉnh sửa 1 cấu trúc dữ liệu sẽ làm thay đổi nội dung hoặc kích cỡ của cấu trúc dữ liệu đó. Như vậy, như đã định nghĩa, `str` và `tuple` không thể chỉnh sửa.

Các thao tác chỉnh sửa chính nên nhớ bao gồm:
* Cập nhật giá trị 1 phần tử
* Thêm hoặc bớt 1 phần tử
* Sắp xếp các phần tử

##### 1.4.1. Cập nhật giá trị 1 phần tử
Các cấu trúc như `list`, `dict` có thể cập nhật giá trị phần tử bằng cách truy cập giá trị của phần tử và gán lại 1 giá trị mới.

```python
print(my_list)
print(my_dict)
my_list[0] = 0
my_dict['one'] = True
print(my_list)
print(my_dict)
```

##### 1.4.2. Thêm phần tử
Cả `list`, `set`, và `dict` đều có thể thêm 1 phần tử mới. Với `list` ta sẽ có thể dùng các phương thức `append()`, `extend()`. Hãy xem thử 2 phương thức này làm những gì:

```python
my_list.append(3)
print(my_list)
```

```python
my_list.append(4, 5)
```

Vậy `append()` chỉ có thể nhận 1 đối số (argument) duy nhất, nên không thể cùng lúc thêm nhiều giá trị. Nếu đối số đó là 1 `tuple`, `list`, `set` hay `dict` thì chính những đối tượng này sẽ được thêm vào, chứ không phải thêm từng phần tử trong những đối tượng đó.

```python
my_list.append(['four', 'five'])
print(my_list)
```

Còn nếu muốn thêm 1 lúc nhiều phần tử, ta có thể dùng `extend()`.

```python
my_list.extend([4, 5])
print(my_list)
```

`extend()` sẽ nhận 1 đối số dạng có thể lặp (iterable), và thêm từng phần tử vào `list`. Như vậy, dù là `list`, `set`, `str` hay `tuple` đều có thể truyền vô `extend()`. Nếu là `dict`, ta có thể thêm các key, các value, hay cả cặp key-value.

```python
six_seven_dict = {'six': 6, 'seven': 7}
my_list.extend(six_seven_dict)
print(my_list)
```

```python
my_list.extend(six_seven_dict.keys())
print(my_list)
```

```python
my_list.extend(six_seven_dict.values())
print(my_list)
```

```python
my_list.extend(six_seven_dict.items())
print(my_list)
```

Các thao tác thêm này sẽ kéo dài, tức là thêm ở cuối `list`. Để chèn tại 1 vị trí bất kì, ta dùng `insert()`. Phương thức này nhận 2 đối số, thứ nhất là vị trí cần chèn, thứ nhì là giá trị chèn.

```python
my_list.insert(1, 1)
print(my_list)
```

Để thêm 1 phần tử vào `set`, ta dùng `add()`. Đối số là giá trị muốn thêm. Tập hợp khi in ra chỉ có sự cập nhật nếu giá trị thêm phân biệt với những giá trị đã có sẵn.

```python
my_set.add(3)
print(my_set)
```

Để cập nhật `dict`, ta đơn giản là dùng cú pháp truy cập và gán cho key mới 1 giá trị,

```python
my_dict['four'] = 4
print(my_dict)
```

Muốn cập nhật nhiều cặp key-value cùng lúc thì dùng `update()`.

```python
my_dict.update(six_seven_dict)
print(my_dict)
```

Nếu gọi `pop()` không cần đối số, sẽ tương đương với `pop(0)`, khi đó `list` sẽ xoá mất phần tử cuối và xuất giá trị của nó.

##### 1.4.4 Sắp xếp phần tử
Kiểu `list` có phương thức `sort()` cho việc sắp xếp các phần tử. Mặc định nếu không truyền tham số nào thì sẽ sắp xếp tăng dần.


```python
list1 = [1, 5, 4, 3]
list1.sort()
list1
```

```python
list2 = ['one', 'two', 'three', 'four']
list2.sort()
list2
```

Dễ thấy mặc định `sort()` sắp xếp tăng dần cả số và kí tự. Tuy nhiên nếu phần tử có nhiều kiểu dữ liệu thì lại khác.

```python
list3 = [1, 'two', 3]
list3.sort()
list3
```

Nếu tiêu chí sắp xếp không rõ ràng, ta buộc phải truyền tham số rõ ràng. Phương thức `sort()` có 2 tham số: `key` và `reverse`. `reverse` thì luôn mặc định là `False`, nghĩa là sắp xếp tăng dần. Vậy nếu ta gán bằng `True` thì sẽ sắp xếp giảm dần. `key` mặc định là `None`, nghĩa là `sort()` sẽ sắp xếp dựa trên chính kiểu dữ liệu của phần tử, nếu được. Vậy nếu không sắp xếp được theo mặc định, ta buộc phải cung cấp `key` cụ thể.

Để quy định cụ thể 1 tiêu chí sắp xếp theo ý mình, ta sẽ sử dụng lamba. Ví dụ, để sắp xếp với `key` là độ dài phần tử, ta làm như sau:

```python
list3.sort(key=lambda x: len(str(x)), reverse=True) # sap xep giam dan
list3
```

`lambda` thực chất là 1 hàm. Trước dấu ":" là input, sau dấu ":" là 1 quy tắc để sinh ra output.

####1.5 Comprehension
Comprehension là 1 thao tác cho phép duyệt qua 1 cấu trúc iterable nào đó, rồi tạo 1 cấu trúc dữ liệu mới từ các phần tử thoả điều kiện nào đó. Có 3 loại comprehension là list comprehension, set comprehension, và dict comprehension.
##### 1.5.1 List comprehension
Giả sử ta có 1 list gồm các số nguyên và cần tính bình phương của chúng.


```python
ints = list(range(10**6))
len(ints)
```

Ta có thể sử dụng vòng lặp. Đo thời gian trung bình của đoạn code trong cả cell với lệnh magic `%%timeit`.

```python
%%timeit
squares = []
for num in ints:
  squares.append(num**2)
```

Dùng list comprehension sẽ gọn hơn về mặt cú pháp vì nó chỉ mất 1 dòng. Đo thời gian trung bình của dòng code với lệnh magic `%timeit`.

```python
%timeit sq = [x * x for x in ints]
```

Như vậy, Python đã cung cấp 1 công cụ thuận tiện và nhanh gọn với list comprehension. Thậm chí, ta hoàn toàn có thể thêm 1 điều kiện để làm những việc phức tạp hơn, như trong ví dụ này, ta hãy chọn những bình phương chẵn không thôi.


```python
%%timeit
even_squares = []
for num in ints:
  if num * num % 2 == 0:
    even_squares.append(num * num)
```

```python
%timeit even_sq = [x * x for x in ints if x * x % 2 == 0]
```

##### 1.5.2 Set comprehension
Tương tự, set comprehension sẽ tạo ra 1 `set` mới sau khi duyệt qua 1 cấu trúc iterable, và thoả điều kiện nào đó.

```python
labels = ["AI", "python", "ai", "Python"]

normalized_labels = {label.lower() for label in labels}
normalized_labels
```

```python
capitalized_labels = {label for label in labels if label[0].isupper()}
capitalized_labels
```

1.5.3. Dictionary comprehension

Tương tự, dict comprehension sẽ tạo ra 1 `dict` mới sau khi duyệt qua 1 iterable, thoả điều kiện nào đó. Cần phải chỉ rõ cặp key-value được hình thành ra sao.



```python
names = ["An", "Binh", "Chi"]

name_lengths = {name: len(name) for name in names}
name_lengths
```

```python
scores = {"An": 7, "Binh": 4, "Chi": 9}

passed_scores = {
    name: score
    for name, score in scores.items()
    if score >= 5
}
passed_scores
```
